## Objective
Create informative visualizations from the Titanic dataset using Matplotlib and Seaborn. The analysis includes distributions, box plots, a correlation matrix, and category comparisons.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

## 1. Load the dataset

```python
df = pd.read_csv('../data/titanic.csv')
df.head()
```

```python
print('Dataset shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nMissing values:')
print(df.isnull().sum())
```

## 2. Histogram — Age Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'].dropna(), bins=20, kde=True)
plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

**Analysis:** The histogram shows the distribution of passenger ages and highlights the concentration of passengers in the younger and middle-age ranges.

## 3. Box Plot — Fare by Passenger Class

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Pclass', y='Fare')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

**Analysis:** Fares differ substantially across passenger classes. Higher-class passengers generally paid higher fares, with some high-value outliers.

## 4. Correlation Matrix Heatmap

```python
corr_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
plt.figure(figsize=(8, 6))
sns.heatmap(df[corr_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Numeric Titanic Variables')
plt.show()
```

**Analysis:** The heatmap summarizes linear relationships among numeric variables. Correlation indicates association, not causation.

## 5. Count Plot — Survival by Passenger Class

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

**Analysis:** The count plot compares survivors and non-survivors within each passenger class and makes differences between classes easy to observe.

## 6. Count Plot — Survival by Gender

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

**Analysis:** The chart compares survival outcomes across genders and provides a clear categorical view of the survival distribution.

## 7. Overall Survival Rate

```python
survival_rate = df['Survived'].mean() * 100
print(f'Overall survival rate: {survival_rate:.2f}%')
```

## Conclusion
This visualization task demonstrates how Matplotlib and Seaborn can be used to explore distributions, identify outliers, examine correlations, and compare categorical groups. The charts provide visual evidence for patterns in passenger age, fare, passenger class, gender, and survival outcomes.
