## Objective
Build and evaluate a supervised machine learning classifier using Scikit-learn. This project uses the Titanic dataset to predict whether a passenger survived.

### Key steps
- Load and inspect the dataset
- Clean missing values and prepare features
- Split data into training and testing sets
- Train a Decision Tree Classifier
- Evaluate the model using accuracy and a confusion matrix
- Review example predictions and summarize results

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 1. Load the Titanic dataset

```python
df = pd.read_csv('../data/titanic.csv')
df.head()
```

```python
print('Dataset shape:', df.shape)
print('\nColumns:')
print(df.columns.tolist())
print('\nMissing values:')
print(df.isnull().sum())
```

## 2. Prepare the features
The target variable is `Survived`. We use passenger class, sex, age, family-related variables and fare as predictors. The passenger name, ticket and cabin are excluded because they are not required for this introductory classifier.

```python
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
target = 'Survived'

model_df = df[features + [target]].copy()

# Fill numeric missing values with the median.
model_df['Age'] = model_df['Age'].fillna(model_df['Age'].median())
model_df['Fare'] = model_df['Fare'].fillna(model_df['Fare'].median())

# Fill the most common embarkation port.
model_df['Embarked'] = model_df['Embarked'].fillna(model_df['Embarked'].mode()[0])

# Convert categorical variables to numeric dummy variables.
X = pd.get_dummies(model_df[features], columns=['Sex', 'Embarked'], drop_first=True, dtype=int)
y = model_df[target].astype(int)

print('Features used:', X.columns.tolist())
print('Missing values after preparation:', X.isnull().sum().sum())
```

## 3. Train/test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print('Training samples:', len(X_train))
print('Testing samples:', len(X_test))
```

## 4. Train a Decision Tree Classifier

```python
model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print('Model training completed.')
```

## 5. Evaluate the model

```python
accuracy = accuracy_score(y_test, y_pred)
print(f'Test Accuracy: {accuracy:.2%}')
print('\nClassification Report:')
print(classification_report(y_test, y_pred, target_names=['Did not survive', 'Survived']))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm, annot=True, fmt='d', cmap='Blues', cbar=False,
    xticklabels=['Did not survive', 'Survived'],
    yticklabels=['Did not survive', 'Survived']
)
plt.title('Confusion Matrix — Decision Tree Classifier')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

## 6. Example predictions

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})
comparison['Actual Label'] = comparison['Actual'].map({0: 'Did not survive', 1: 'Survived'})
comparison['Predicted Label'] = comparison['Predicted'].map({0: 'Did not survive', 1: 'Survived'})
comparison.head(10)
```

## 7. Conclusion
The Decision Tree classifier demonstrates a complete supervised machine learning workflow: data preparation, train/test splitting, model training, prediction, and evaluation. Accuracy and the confusion matrix provide a simple view of how well the classifier performs on unseen test data.

The result is intended for learning the machine learning pipeline; it should not be interpreted as a historical causal explanation of Titanic survival.
